```python
import pandas as  pd
df = pd.read_excel(r"C:\Users\Computer\Downloads\retail_outlier_practice.xlsx")
df
```

```python
Q1 = df["unit_price"].quantile(0.25)
Q3 = df["unit_price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["price_outlier"] = (
    (df["unit_price"] < lower_bound) |
    (df["unit_price"] > upper_bound)
)
```

```python
df.loc[df["price_outlier"],["order_id","product_name","unit_price","quantity"]]
```

```python
df["review_status"] = "Normal"

df.loc[df["price_outlier"], "review_status"] = "Needs review"
```
