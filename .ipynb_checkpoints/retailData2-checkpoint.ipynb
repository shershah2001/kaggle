```python
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("umerkk12/online-retail-business")
df = pd.read_csv(
    r"C:\Users\Computer\.cache\kagglehub\datasets\umerkk12\online-retail-business\versions\1\OnlineRetail.csv",encoding="latin-1")
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"],format="%d-%m-%Y %H:%M")
# df["InvoiceNo"].map(type).value_counts()
df["InvoiceNO"] = df["InvoiceNo"].astype("str")
# df["StockCode"].map(type).value_counts()
df["StockCode"]=df["StockCode"].astype("str")
# df["StockCode"].map(type).value_counts()
df["InvoiceNo"].str[0].value_counts()
df["InvoiceNo"].str.startswith("C").value_counts()
df = df[~df["InvoiceNo"].str.startswith("C")]
df["InvoiceNo"].str[0].value_counts()
df.describe()
df=df[(df["Quantity"]> 0) & (df["UnitPrice"]> 0)]

df["Description"] = df["Description"].str.strip()
df["Country"] = df["Country"].str.strip()
# pd.set_option("display.max_rows", None)

df = df[df["StockCode"].str.contains(r"\d+",regex=True)]
df["StockCode"].value_counts()
df["InvoiceNo"].value_counts()
df.duplicated().sum()
df.duplicated().value_counts()
df.drop_duplicates()
df["CustomerID"].isnull().sum()
df["CustomerID"] = df["CustomerID"].astype("Int64")
# 1) Sales/revenue analysis ke liye - full data
df_sales = df.copy()
# 2) Customer analysis (RFM, segmentation) ke liye - sirf valid CustomerID
df_customer = df.dropna(subset=["CustomerID"])
df_customer["CustomerID"].isnull().sum()
df_customer["TotalPrice"] = df_customer["Quantity"] * df_customer["UnitPrice"]
df_customer["Month"] = df_customer["InvoiceDate"].dt.month
df_customer["Hour"] = df_customer["InvoiceDate"].dt.hour
df_customer.reset_index(drop=True).to_csv("OnlineRetail_clean.csv", index=False)
df_customer
```
