# Replacing Month Names Using Regex and Dictionary

```python
month_map = {
    "Jan": "01", "Feb": "02", "Mar": "03",
    "Apr": "04", "May": "05", "Jun": "06",
    "Jul": "07", "Aug": "08", "Sep": "09",
    "Oct": "10", "Nov": "11", "Dec": "12",

    "January": "01", "February": "02", "March": "03",
    "April": "04", "June": "06", "July": "07",
    "August": "08", "September": "09", "October": "10",
    "November": "11", "December": "12"
}

pattern = r"\b(" + "|".join(month_map.keys()) + r")\b"

df["order_date"] = df["order_date"].str.replace(
    pattern,
    lambda m: month_map[m.group()],
    regex=True
)
pattern=r'^\d{4}-\d{2}-\d{2}$'
pattern1=r'[a-zA-Z]'
pattern2=r'[@#$%^&*()_]'
df.loc[~df["order_date"].str.contains(pattern,na=False),"order_date"].unique()
```

```python
domain = ["gmail.com", "yahoo.com", "outlook.com", "rediffmail.com"]
def fix_email(email):
    if pd.isna(email):
        return email

    if "@" in email:
        return email

    for i in domain:
        if email.endswith(i):
            index = email.find(i)
            return email[:index] + "@" + email[index:]

    return email

df["email"] = df["email"].apply(fix_email)
```

```python
import numpy as np
base = df['quantity'] * df['unit_price']
implied = (1 - df['total_amount'] / base).round(2)
invalid = ~df['discount_pct'].isin([0,5,10,15,20,25,30]) & df['discount_pct'].notna()
fix = invalid & implied.isin([0,.05,.10,.15,.20,.25,.30])
df.loc[fix, 'discount_pct'] = implied[fix] * 100
df.loc[invalid & ~fix, 'discount_pct'] = np.nan
df["discount_pct"]
```

```python
price_ref = (df.loc[df['unit_price']>1]
               .groupby('product_name')['unit_price']
               .agg(lambda s: s.mode().iloc[0]))
df['unit_price'] = df['unit_price'].fillna(df['product_name'].map(price_ref))
df.loc[df['unit_price']<=1,'unit_price'] = np.nan
# df['unit_price'] = df['unit_price'].fillna(df['product_name'].map(price_ref))
df['total_amount'] = df['total_amount'].abs()
valid = [0,5,10,15,20,25,30]
base = df['quantity'] * df['unit_price']
implied = (1 - df['total_amount'] / base).round(2)
invalid = ~df['discount_pct'].isin(valid) & df['discount_pct'].notna()
fix = invalid & implied.isin([v/100 for v in valid])
df['discount_fixed'] = fix  
calc = df['quantity']*df['unit_price']*(1-df['discount_pct']/100)
bad = df['total_amount'].notna() & calc.notna() & ((df['total_amount']-calc).abs()>1)
df.loc[bad,'total_amount'] = calc[bad]
df['total_amount'] = df['total_amount'].fillna(calc)
```

# 14 Customer retention rate / churn rate kya hai?

```python
first_purchase = df.groupby("customer_id")["order_date"].min()
first_purchase 
df["order_date"] = pd.to_datetime(df["order_date"],format="%Y-%m-%d")
df["month"] = df["order_date"].dt.to_period("M")
monthly_customers = df.groupby("month")["customer_id"].apply(
    lambda x: set(x.unique())
)
monthy_customer
results =[]
for i in range(0, len(monthly_customers)-1):
    start_customers = monthly_customers.iloc[i]
    end_customers = monthly_customers.iloc[i+1]
    retained = start_customers.intersection(end_customers)

    S = len(start_customers)
    R = len(retained)

    retention = (R / S) * 100 if S else 0
    churn = ((S - R) / S) * 100 if S else 0
    results.append({
        "month": monthly_customers.index[i],
        "starting_customers": S,
        "retained_customers": R,
        "retention_rate": retention,
        "churn_rate": churn
    })

retention_df = pd.DataFrame(results)
retention_df

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    retention_df["month"].astype(str),
    retention_df["retention_rate"],
    linestyle="dotted",
    c="green",
    linewidth=2.5,
    marker="o",
    label="Retention Rate"
)

plt.plot(
    retention_df["month"].astype(str),
    retention_df["churn_rate"],
    linestyle="solid",
    c="red",
    linewidth=2.5,
    marker="o",
    label="Churn Rate"
)

plt.xlabel("Month", size=12)
plt.ylabel("Percentage (%)", size=12)
plt.title("Monthly Customer Retention and Churn Rate", pad=20)

plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

# 15. RFM segmentation karo – Champions, Loyal, At-risk, Lost customers kaun hain

```python
print(rfm.shape)

print(rfm.isnull().sum())

print(rfm[["recency", "Frequency", "Monetary"]].describe())
```

```python
missing_customers = df.groupby("customer_id")["order_date"].apply(
    lambda x: x.isna().all()
)

print("Customers with all dates missing:", missing_customers.sum())
```

```python
rfm = rfm.dropna(subset=["recency"]).copy()
```

```python
rfm.isna().sum()
```

```python
import pandas as pd

# Step 1: Convert order_date into datetime
df["order_date"] = pd.to_datetime( 
    df["order_date"],
    format="%Y-%m-%d"
)

# Step 2: Reference date
Reference_date = df["order_date"].max() + pd.Timedelta(days=1)

# Step 3: Create RFM table
rfm = df.groupby("customer_id").agg(
    recency=("order_date", lambda x: (Reference_date - x.max()).days),
    Frequency=("transaction_id", "nunique"),
    Monetary=("total_amount", "sum")
)

# Step 4: Remove customers whose Recency is missing
rfm = rfm.dropna(subset=["recency"]).copy()

# Step 5: RFM scoring
rfm["R_score"] = pd.qcut(
    rfm["recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Step 6: Customer segmentation
def customer_segment(row):

    if (row["R_score"] >= 4 and
        row["F_score"] >= 4 and
        row["M_score"] >= 4):
        return "Champions"

    elif row["R_score"] >= 3 and row["F_score"] >= 4:
        return "Loyal Customers"

    elif row["R_score"] <= 2 and row["F_score"] >= 3:
        return "At-risk Customers"

    elif row["R_score"] <= 2 and row["F_score"] <= 2:
        return "Lost Customers"

    else:
        return "Others"

# Step 7: Apply segmentation
rfm["Segment"] = rfm.apply(customer_segment, axis=1)

# Step 8: Display result
rfm.head(10)
```

```python
import matplotlib.pyplot as plt

segment_count = rfm["Segment"].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(segment_count.index, segment_count.values)
# print(segment_count.values)
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.title("Customer Segmentation Analysis")

plt.xticks(rotation=45)
plt.show()
```

# 19. Pehli khareed ke baad customer wapas kitne din mein aata hai?

```python
df["order_date"] = pd.to_datetime(df["order_date"])

# Har customer ki purchase dates sort karo
purchase = df.sort_values(["customer_id", "order_date"])

# Customer ki previous purchase date nikalo
purchase["previous_purchase"] = (
    purchase.groupby("customer_id")["order_date"].shift(1)
)

# Do purchases ke beech kitne din ka gap hai
purchase["days_between"] = (
    purchase["order_date"] - purchase["previous_purchase"]
).dt.days

# Har customer ki pehli repeat purchase ka gap
repeat_purchase = (
    purchase.dropna(subset=["days_between"])
    .groupby("customer_id")["days_between"]
    .first()
)

print("Average days to repeat purchase:", repeat_purchase.mean())
```

# 41. Pareto (80/20): Kitne % customers/products 80% revenue dete hain

```python
import pandas as pd
import matplotlib.pyplot as plt

# Customer-wise revenue calculate karo
customer_revenue = (
    df.groupby("customer_id")["total_amount"]
    .sum()
    .sort_values(ascending=False)
)

# Cumulative revenue calculate karo
cumulative_revenue = customer_revenue.cumsum()

# Cumulative revenue percentage
cumulative_percentage = (
    cumulative_revenue / customer_revenue.sum()
) * 100

# 80% revenue tak kitne customers hain?
customers_80 = (cumulative_percentage < 80).sum() + 1

# Total customers ke comparison mein percentage
customer_percentage = (
    customers_80 / len(customer_revenue)
) * 100

print("Customers required for 80% revenue:", customers_80)
print("Percentage of customers:", round(customer_percentage, 2), "%")
```

# 42. Next 3 months ka sales forecast kya hoga?

```python
import pandas as pd

df["order_date"] = pd.to_datetime(df["order_date"])

monthly_sales = (
    df.set_index("order_date")["total_amount"]
    .resample("MS")
    .sum()
)

print(monthly_sales)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.arange(len(monthly_sales)).reshape(-1, 1)

y = monthly_sales.values

model = LinearRegression()

model.fit(X, y)
```

```python
future_months = np.arange(
    len(monthly_sales),
    len(monthly_sales) + 3
).reshape(-1, 1)

forecast = model.predict(future_months)
```

```python
future_dates = pd.date_range(
    start=monthly_sales.index[-1] + pd.offsets.MonthBegin(1),
    periods=3,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Month": future_dates,
    "Predicted_Sales": forecast.round(2)
})

print(forecast_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

# Actual sales
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    label="Actual Sales"
)

# Future predicted sales
plt.plot(
    future_dates,
    forecast,
    marker="o",
    linestyle="--",
    color="red",
    label="Predicted Sales"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Next 3 Months Sales Forecast")

plt.legend()
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

# 43. Kaunse customers churn hone wale hain? (simple model)¶

```python
import pandas as pd

df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

customer_data = df.groupby("customer_id").agg(
    last_purchase=("order_date", "max"),
    total_spending=("total_amount", "sum"),
    total_orders=("transaction_id", "nunique")
)

customer_data.head()
```

```python
reference_date = df["order_date"].max()

customer_data["days_inactive"] = (
    reference_date - customer_data["last_purchase"]
).dt.days
```

```python
def churn_status(days):
    if days >= 180:
        return "High Risk"
    elif days >= 90:
        return "Medium Risk"
    else:
        return "Active"

customer_data["churn_status"] = (
    customer_data["days_inactive"].apply(churn_status)
)

customer_data.head(10)
```

```python
churn_count = customer_data["churn_status"].value_counts()

print(churn_count)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    churn_count.index,
    churn_count.values,
    color=["green", "orange", "red"]
)

plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.title("Customer Churn Risk Analysis")

plt.show()
```

```python
high_risk_customers = customer_data[
    customer_data["churn_status"] == "High Risk"
].sort_values("days_inactive", ascending=False)

high_risk_customers.head(20)
```
